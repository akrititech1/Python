```python
#1 Create a DataFrame from a dictionary
import pandas as pd
data={
    "SID":[101,102,103,104,105],
    "Name":["Bob","Jack","Rio","Tom","Sia"],
    "City":["Delhi","Kolkata","Mumbai","Pune","Chennai"],
    "Course":["IT","MECH","ELEC","CIVIL","CHEM"],
    "Marks":[67,45,89,58,90],
    "Grade":["B","C","A","B","A+"]
}
df=pd.DataFrame(data)
print(df)
```

```python
#2 To remove duplicate rows
import pandas as pd
data={
    "SID":[101,102,101,104,105],
    "Name":["Bob","Jack","Bob","Tom","Sia"],
    "City":["Delhi","Kolkata","Delhi","Pune","Chennai"],
    "Course":["IT","MECH","IT","CIVIL","CHEM"],
    "Marks":[67,45,67,58,90],
    "Grade":["B","C","B","B","A+"]
}
df=pd.DataFrame(data)
print(df)
print("==================Removing Duplicates====================================")
# df[df.duplicated(keep=False)]
df.drop_duplicates(inplace=True)
print(df)
```

```python
#3 To remove a columns from a dataframe
import pandas as pd
data={
    "SID":[101,102,103,104,105],
    "Name":["Bob","Jack","Rio","Tom","Sia"],
    "City":["Delhi","Kolkata","Mumbai","Pune","Chennai"],
    "Course":["IT","MECH","ELEC","CIVIL","CHEM"],
    "Marks":[67,45,89,58,90],
    "Grade":["B","C","A","B","A+"]
}
df=pd.DataFrame(data)
print(df)
df=df.drop(["City"],axis=1)                        #Method 1
print(df)
df=df.drop(columns="Name")                         #Method 2
print(df)
df=df.drop(["Course","Grade"],axis=1)              #Method 3
print(df)
```

```python
#4 To perform Aggregation using GroupBy
import pandas as pd
data={
    "SID":[101,102,103,104,105],
    "Name":["Bob","Jack","Rio","Tom","Sia"],
    "City":["Delhi","Kolkata","Mumbai","Pune","Chennai"],
    "Course":["IT","MECH","ELEC","IT","MECH"],
    "Marks":[67,45,89,58,90],
    "Grade":["B","C","A","B","A+"]
}
df=pd.DataFrame(data)
print(df)
result=df.groupby("Course")["Marks"].mean()
print(result)

```

```python
#5 To find Average Salary of Emp by using Group by JOB
import pandas as pd
data ={
    "EID":[101,102,103,104,105],
    "Name":["Bob","Jack","Rio","Tom","Sam"],
    "Department":["IT","Finance","Telecom","HR","IT"],
    "Salary":[34000,58000,23000,78000,40000]
}
df=pd.DataFrame(data)
print(df)
result=df.groupby("Department")["Salary"].mean()
print(result)
```

```python
#9  How to Apply a Function to Each Element in a column
import pandas as pd
data ={
    "EID":[101,102,103,104,105],
    "Name":["Bob","Jack","Rio","Tom","Sam"],
    "Department":["IT","Finance","Telecom","HR","IT"],
    "Salary":[34000,58000,23000,78000,40000]
}
df=pd.DataFrame(data)
print(df)
df["New_Salary"]=df["Salary"].apply(lambda x:x+500)
print(df)

```

```python
#10 How to Merge Two DataFrames
import pandas as pd
data1={
    "SID":[101,102,103],
    "Name":["Bob","Jack","Rio"],
    "Country":["US","India","Japan"]
}
df1=pd.DataFrame(data1)
print(df1)

data2={
    "SID":[101,102,103],
    "M1":[34,56,78],
    "M2":[45,89,98],
    "M3":[64,51,60],
}
df2=pd.DataFrame(data2)
print(df2)

df=pd.merge(df1,df2,on="SID")
print(df)
```

```python
#11  How to Pivot a DataFrame
import pandas as pd
data ={
    "EID":[101,102,103,104,105],
    "Name":["Bob","Jack","Rio","Tom","Sam"],
    "Department":["IT","Finance","Telecom","HR","IT"],
    "Salary":[34000,58000,23000,78000,40000]
}
df=pd.DataFrame(data)
print(df)
print("========================Pivot=============================")
pivot_df=df.pivot(index="EID",columns="Name",values="Department")
print(pivot_df)
```

```python
#12 Filter and Pivot Functions
import pandas as pd
data ={
    "EID":[101,102,103,104,105],
    "Name":["Bob","Jack","Rio","Tom","Sam"],
    "Department":["IT","Finance","Telecom","HR","IT"],
    "Salary":[34000,58000,23000,78000,40000]
}
df=pd.DataFrame(data)
print("=====================Original Data============================")
print(df)
print("=====================Filter Data============================")
filter_df=df.filter(items=["EID","Name"])
print(filter_df)
print("=====================Pivot Data============================")
pivot_df=df.pivot(index="Department",columns="Name",values="Salary")
print(pivot_df)
```

```python
#13 How to Handle Missing Values in a DataFrame
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
data ={
    "EID":[101,102,103,104,105],
    "Name":["Bob",np.nan,"Rio",np.nan,"Sam"],
    "Department":["IT","Finance","Telecom","HR",np.nan],
    "Salary":[34000,58000,np.nan,78000,40000]
}
df=pd.DataFrame(data)
print(df)
# df.dropna(inplace=True)
# print(df)
df["Name"].fillna("Jack",inplace=True)
df["Department"].fillna("IT",inplace=True)
print(df)
```

```python
#14 How to Perform Arithmetic Operations on Columns:
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

data={
    "SNO" : [101,102,103,104,102],
    "SNAME" : ['RAM','JOHN','MILLER','MAHESH','JOHN'],
    "M1" : [90,45,25,56,45],
    "M2" : [45,37,56,89,37],
    "M3" : [67,47,59,45,47],
    "VIVA" : [10,8,np.nan,9,8]
}
#to create data frame
df=pd.DataFrame(data)
print(df)
#print(" ---------------------------------")
#df.info()
#print(" ---------------------------------")
#to remove duplicate rows
df.drop_duplicates(inplace=True)
#to replace Null values
df.VIVA.fillna(0,inplace=True)
print(" ========================================")
print(df)
#to find total_marks
def total(df):
   return df['M1']+df['M2']+df['M3']+df['VIVA']

df['TOTAL_MARKS']=df.apply(total,axis=1)

#to find Result
def result(df):
   if (df['M1']>34 and df['M2']>34 and df['M3']>34):
        return 'PASS'
   else:
        return 'FAIL'
df['RESULT']=df.apply(result,axis=1)
print(" ========================================")

print(df)
print(" ========================================")
No_of_Students=df.SNO.count()
Grand_Total_Marks = df.TOTAL_MARKS.sum()
Grand_Viva_Marks = df.VIVA.sum()
print(" Total No.of Stuents : ",No_of_Students)
print(" Grand Total Marks : ",Grand_Total_Marks)
print(" Grand VIVA Marks : ",Grand_Viva_Marks)
print(" ========================================")

```
