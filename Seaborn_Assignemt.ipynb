```python
#Q13 Create a heatmap of the correlation matrix for the penguins dataset (numeric columns only).

import matplotlib.pyplot as plt
import seaborn as sns
penguins=sns.load_dataset("penguins")
print(penguins.head(5))
numeric_penguins = penguins.select_dtypes(include="number")
#penguins → Original DataFrame.
#select_dtypes() → Selects columns based on their data type.
#include="number" → Keeps only numeric columns (int, float, etc.).
#numeric_penguins → A new DataFrame containing only numeric columns.
corr = numeric_penguins.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
```

```python
#Q28 Write a complete script that combines at least 4 different Seaborn plots in a 2x2
#subplot grid using Matplotlib subplots.
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
sns.histplot(data=tips, x="total_bill", kde=True, ax=axes[0,0])
sns.boxplot(data=tips, x="day", y="total_bill", ax=axes[0,1])
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex", ax=axes[1,0])
sns.heatmap(tips.select_dtypes(include="number").corr(), annot=True, ax=axes[1,1])
plt.tight_layout()
plt.show()
```

```python
#Q30 What are Seaborn themes? Apply 'darkgrid' and demonstrate with a line plot.

# Seaborn has 5 built-in themes: darkgrid, whitegrid, dark, white, ticks
# Apply with sns.set_theme() or sns.set_style()

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_theme(style="darkgrid",palette="Reds",font_scale=1.1)

x=np.linspace(0,10,100)
y=np.sin(x)

sns.lineplot(x=x,y=y,color="Purple",linewidth=2.5)
plt.xlabel("X")
plt.ylabel("Sin X")
plt.title("Seaborn with darkgrid theme")
plt.show()
```
