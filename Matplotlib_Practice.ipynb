```python
#Q39 Write a script to plot a bar chart for monthly sales.
import matplotlib.pyplot as plt
import numpy as np
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [500, 750, 600, 900]
colors = ['hotpink', 'blue', 'yellow', 'red']
#plt.bar(months, sales, color=colors, width=0.5, edgecolor='black')
plt.figure(figsize=(8, 5))
bars = plt.bar(months, sales, color=colors, width=0.5, edgecolor='black')
# Add value labels on top of bars
for bar in bars:
    plt.text(
    bar.get_x() + bar.get_width()/2,
    bar.get_height() + 10,
    str(bar.get_height()),
    ha='center',
    va='bottom',
    fontweight='bold'
    )
plt.title("Monthly Sales Report", fontsize=14, fontweight='bold')
plt.xlabel("Month")
plt.ylabel("Sales (USD)")
plt.grid(axis='y', alpha=0.5)
#plt.tight_layout()
plt.show()
```

```python
#Q40 How do you plot multiple lines on the same graph?
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
plt.figure(figsize=(10, 5))
plt.plot(x, y1, label='sin(x)', color='blue', linestyle='-')
plt.plot(x, y2, label='cos(x)', color='red', linestyle='--')
plt.plot(x, y3, label='tan(x)', color='green', linestyle=':')
plt.ylim(-2, 2)
plt.title("Multiple Lines on Same Graph")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Q41 Write a Python script to create a scatter plot for 20 random data points.
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
x = np.random.randn(20)
y = np.random.randn(20)
colors = np.random.rand(20)
sizes = np.random.randint(50, 300, 20)
plt.figure(figsize=(8, 6))
scatter = plt.scatter(x, y, c=colors, s=sizes,
cmap='viridis', alpha=0.8, edgecolors='black')
plt.colorbar(scatter, label='Color Value')
plt.title("Scatter Plot – 20 Random Points", fontsize=14)
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#Q42 Write a script with a 2×2 grid of subplots.
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 100)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# First subplot
axes[0,0].plot(x, np.sin(x), 'b-')
axes[0,0].set_title("Sine Wave")

# Second subplot
axes[0,1].plot(x, np.cos(x), 'r--')
axes[0,1].set_title("Cosine Wave")

# Third subplot
axes[1,0].plot(x, np.tan(x), 'g-')
axes[1,0].set_ylim(-5, 5)
axes[1,0].set_title("Tangent Wave")

# Fourth subplot
axes[1,1].plot(x, np.sin(x) * np.cos(x), 'm:')
axes[1,1].set_title("sin(x) · cos(x)")

# Add grid to all subplots
for ax in axes.flat:
    ax.grid(True, alpha=0.3)

plt.suptitle("Trigonometric Functions", fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
#Q43 plot horizontal bar graph using barh.

import matplotlib.pyplot as plt
languages = ['Python', 'JavaScript', 'Java', 'C++', 'Go', 'Rust']
popularity = [85, 75, 65, 55, 40, 30]
colors = ['#4472C4','#ED7D31','#FFC000','#70AD47','#FF0000','#7030A0']
plt.figure(figsize=(9, 5))
bars = plt.barh(languages, popularity, color=colors,
edgecolor='black', height=0.6)
# Add value labels
for bar, val in zip(bars, popularity):
     plt.text(val + 0.5, bar.get_y() + bar.get_height()/2,
     f'{val}%', va='center', fontweight='bold')
     plt.xlim(0, 100)
plt.title("Programming Language Popularity (%)", fontsize=14,
fontweight='bold')
plt.xlabel("Popularity (%)")
plt.grid(axis='x', alpha=0.4)
plt.tight_layout()
plt.show()
```
